```python
import numpy as np
import scipy.stats as stat
from collections import defaultdict 
import pandas as pd
import os, random, sys
```

```python
random.choice([0,1])
```

```python
# input genes
input_genes = pd.read_csv('../../../gene/MSK-IMPACT468.txt', sep='\t')
input_genes = input_genes['gene'].tolist()
```

```python
sampleIDs = np.arange(20)
```

```python
# mut, cna, cnd
for alt_type in ['mut', 'cna', 'cnd']:
    out = defaultdict(list)
    out['sample'] = sampleIDs
    for gidx, gene in enumerate(input_genes):
        rand_ints = random.choices([0,1], weights=[0.8, 0.2], k=len(sampleIDs))
        out[gene] = rand_ints
    out = pd.DataFrame(data=out, columns=np.append(['sample'], input_genes))
    out.to_csv(f'./{alt_type}.txt', sep='\t', index=False)
```

```python
# covariates
out = defaultdict(list)
out['sample'] = sampleIDs
columns = ['sample', 'TMB', 'aSCNA', 'APOBEC', 'Clock_SBS1', 'Clock_SBS5', 'MMR', 'POLE', 'Tobacco', 'UV']

for cov in columns[1:]:
    if cov == 'TMB':
        rand_nums = np.random.uniform(0, 100, size=len(sampleIDs))
    elif cov == 'aSCNA':
        rand_nums = np.random.uniform(0, 1, size=len(sampleIDs))
    else:
        rand_nums = random.choices([0,1], weights=[0.8, 0.2], k=len(sampleIDs))
    out[cov] = rand_nums
out = pd.DataFrame(data=out, columns=columns)
out.to_csv(f'./covariates.txt', sep='\t', index=False)
```

```python
# outcomes
out = defaultdict(list)
out['sample'] = sampleIDs
rand_nums = random.choices([0,1], weights=[0.8, 0.2], k=len(sampleIDs))
out['outcomes'] = rand_nums
out = pd.DataFrame(data=out, columns=['sample', 'outcomes'])
out.to_csv(f'./outcomes.txt', sep='\t', index=False)
```
